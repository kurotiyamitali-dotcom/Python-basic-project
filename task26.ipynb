```python
from sklearn.datasets import make_blobs
import pandas as pd

X, y = make_blobs(n_samples=500, centers=4, cluster_std=0.70)

df = pd.DataFrame(X, columns=['Feature1', 'Feature2'])

print(df.head(10))
print(df.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[['Feature1', 'Feature2']])

print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print(df.head(10))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x='Feature1', y='Feature2', hue='Cluster')

plt.title('K-Means Clusters')
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('K')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

print(df['Cluster'].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(data=df, x='Feature1', y='Feature2', hue='Cluster')

plt.title('K-Means Clustering Results')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print(df.head())
print(df.shape)

X_scaled = StandardScaler().fit_transform(df)

inertia = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('K')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()

kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

plt.scatter(df.iloc[:, 0], df.iloc[:, 1], c=df['Cluster'])
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering Results')
plt.show()
```
