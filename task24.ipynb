```python
import pandas as pd
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

iris = load_iris()

X_classification = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
y_classification = iris.target

Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_classification,
    y_classification,
    test_size=0.2,
    random_state=42
)

scaler_c = StandardScaler()
Xc_train = scaler_c.fit_transform(Xc_train)
Xc_test = scaler_c.transform(Xc_test)

print("Classification X:", X_classification.columns.tolist())
print("Classification y: target")


diabetes = load_diabetes()

X_regression = pd.DataFrame(
    diabetes.data,
    columns=diabetes.feature_names
)
y_regression = diabetes.target

Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X_regression,
    y_regression,
    test_size=0.2,
    random_state=42
)

scaler_r = StandardScaler()
Xr_train = scaler_r.fit_transform(Xr_train)
Xr_test = scaler_r.transform(Xr_test)

print("Regression X:", X_regression.columns.tolist())
print("Regression y: target")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "SVM": SVC(),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB()
}

results = []

for name, model in models.items():

    model.fit(Xc_train, yc_train)
    y_pred = model.predict(Xc_test)

    accuracy = accuracy_score(yc_test, y_pred)

    print("\n", name)
    print("Accuracy:", accuracy)
    print("Confusion Matrix:")
    print(confusion_matrix(yc_test, y_pred))
    print("Classification Report:")
    print(classification_report(yc_test, y_pred))

    results.append([name, accuracy])

comparison = pd.DataFrame(
    results,
    columns=["Model", "Accuracy"]
)

print("\nComparison Table:")
print(comparison)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor

from sklearn.metrics import r2_score

models_reg = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "SVR": SVR(),
    "KNN Regressor": KNeighborsRegressor()
}

results_reg = []

for name, model in models_reg.items():

    model.fit(Xr_train, yr_train)
    y_pred = model.predict(Xr_test)

    r2 = r2_score(yr_test, y_pred)

    print(name, "R² Score:", r2)

    results_reg.append([name, r2])

comparison_reg = pd.DataFrame(
    results_reg,
    columns=["Model", "R2 Score"]
)

print("\nComparison Table:")
print(comparison_reg)
```

```python
import joblib

for name, model in models.items():
    joblib.dump(model, name.replace(" ", "_") + ".pkl")

for name, model in models_reg.items():
    joblib.dump(model, name.replace(" ", "_") + ".pkl")

best_classification = comparison.loc[
    comparison["Accuracy"].idxmax(), "Model"
]

best_regression = comparison_reg.loc[
    comparison_reg["R2 Score"].idxmax(), "Model"
]

joblib.dump(
    models[best_classification],
    "best_classification_model.pkl"
)

joblib.dump(
    models_reg[best_regression],
    "best_regression_model.pkl"
)

joblib.dump(scaler_c, "classification_scaler.pkl")
joblib.dump(scaler_r, "regression_scaler.pkl")

joblib.dump(
    list(X_classification.columns),
    "classification_columns.pkl"
)

joblib.dump(
    list(X_regression.columns),
    "regression_columns.pkl"
)

print("Best models saved successfully.")
```
