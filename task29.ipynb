```python
import pandas as pd

from google.colab import files

uploaded = files.upload()

from sklearn.model_selection import train_test_split

df = pd.read_csv(
    '/content/train.txt',
    sep=';',
    names=['text', 'emotions']
)

df = df.dropna()

X = df['text']
y = df['emotions']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

X_train_bow = vectorizer.fit_transform(X_train)

X_test_bow = vectorizer.transform(X_test)

print("X_train_bow shape:", X_train_bow.shape)
print("X_test_bow shape:", X_test_bow.shape)

print("\nFirst 20 feature names:")
print(vectorizer.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

nb_model = MultinomialNB()

nb_model.fit(X_train_bow, y_train)

y_pred_bow = nb_model.predict(X_test_bow)

bow_accuracy = accuracy_score(y_test, y_pred_bow)

print("Bag of Words Accuracy:", bow_accuracy)
```

```python
vocabulary_size = len(vectorizer.vocabulary_)

print("Total vocabulary size:", vocabulary_size)

print("\n15 words from vocabulary:")
print(list(vectorizer.vocabulary_.keys())[:15])

sample_document = X_train.iloc[0]

print("\nSample document:")
print(sample_document)

sample_vector = vectorizer.transform([sample_document])

print("\nBoW vector:")
print(sample_vector)

print("\nDense BoW vector:")
print(sample_vector.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

bigram_vectorizer = CountVectorizer(ngram_range=(1, 2))

X_train_bigram = bigram_vectorizer.fit_transform(X_train)

X_test_bigram = bigram_vectorizer.transform(X_test)

print("X_train_bigram shape:", X_train_bigram.shape)
print("X_test_bigram shape:", X_test_bigram.shape)

features = bigram_vectorizer.get_feature_names_out()

print("\nSome bigram features:")
print(features[:30])
```

```python
bigram_model = MultinomialNB()

bigram_model.fit(X_train_bigram, y_train)

y_pred_bigram = bigram_model.predict(X_test_bigram)

bigram_accuracy = accuracy_score(y_test, y_pred_bigram)

print("Unigram BoW Accuracy:", bow_accuracy)
print("Unigram + Bigram Accuracy:", bigram_accuracy)

if bigram_accuracy > bow_accuracy:
    print("\nUnigram + Bigram performed better.")
elif bigram_accuracy < bow_accuracy:
    print("\nUnigram BoW performed better.")
else:
    print("\nBoth methods performed equally.")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("X_train_tfidf shape:", X_train_tfidf.shape)
print("X_test_tfidf shape:", X_test_tfidf.shape)

print("\nFirst 15 TF-IDF feature names:")
print(tfidf_vectorizer.get_feature_names_out()[:15])
```

```python
tfidf_model = MultinomialNB()

tfidf_model.fit(X_train_tfidf, y_train)

y_pred_tfidf = tfidf_model.predict(X_test_tfidf)

tfidf_accuracy = accuracy_score(y_test, y_pred_tfidf)

print("TF-IDF Accuracy:", tfidf_accuracy)
```

```python
import pandas as pd

comparison = pd.DataFrame({
    "Method": [
        "Bag of Words (Unigrams)",
        "Bag of Words (Unigrams + Bigrams)",
        "TF-IDF"
    ],
    "Accuracy": [
        bow_accuracy,
        bigram_accuracy,
        tfidf_accuracy
    ]
})

print(comparison)

best_method = comparison.loc[comparison["Accuracy"].idxmax(), "Method"]
best_accuracy = comparison["Accuracy"].max()

print("\nBest Method:", best_method)
print("Best Accuracy:", best_accuracy)

print("\nObservation:")
print(
    f"{best_method} performed best with an accuracy of "
    f"{best_accuracy:.4f}. It performed better because its "
    "vectorization method represented the text more effectively "
    "for emotion classification."
)
```

```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

from google.colab import files

uploaded = files.upload()

df = pd.read_csv(
    '/content/train.txt',
    sep=';',
    names=['text', 'emotions']
)

df = df.dropna()

X = df['text']
y = df['emotions']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

count_vectorizer = CountVectorizer()

X_train_count = count_vectorizer.fit_transform(X_train)
X_test_count = count_vectorizer.transform(X_test)

count_model = MultinomialNB()
count_model.fit(X_train_count, y_train)

count_pred = count_model.predict(X_test_count)
count_accuracy = accuracy_score(y_test, count_pred)


tfidf_vectorizer = TfidfVectorizer()

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

tfidf_model = MultinomialNB()
tfidf_model.fit(X_train_tfidf, y_train)

tfidf_pred = tfidf_model.predict(X_test_tfidf)
tfidf_accuracy = accuracy_score(y_test, tfidf_pred)


print("CountVectorizer Accuracy:", count_accuracy)
print("TF-IDF Accuracy:", tfidf_accuracy)


if count_accuracy >= tfidf_accuracy:
    best_vectorizer = count_vectorizer
    best_model = count_model
else:
    best_vectorizer = tfidf_vectorizer
    best_model = tfidf_model

joblib.dump(best_vectorizer, 'best_vectorizer.pkl')
joblib.dump(best_model, 'best_model.pkl')

print("Best vectorizer saved successfully.")
print("Best model saved successfully.")
```
